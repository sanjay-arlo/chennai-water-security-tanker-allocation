# Chennai Water Security & Tanker Allocation

Portfolio analysis notebook specification. The checked-in CSV is synthetic; external 2026 context is stored separately in `data/source_snapshot.csv`.


```python
import pandas as pd
df = pd.read_csv('sample_reservoir_zone_data_chennai.csv', parse_dates=['date'])
df.head()
```

```python
profile = (df.groupby('zone')
           .agg(avg_storage_pct=('storage_pct','mean'),
                avg_demand_mld=('daily_demand_mld','mean'),
                avg_tanker_supply_mld=('tanker_supply_mld','mean'),
                avg_rainfall_deficit_pct=('rainfall_deficit_pct','mean'))
           .sort_values('avg_storage_pct'))
profile
```

```python
monthly = df.groupby('date').agg(avg_storage_pct=('storage_pct','mean'), avg_tanker_supply_mld=('tanker_supply_mld','mean'))
monthly
```
